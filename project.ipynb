#### Exercise 10. Are the average prices per square meter (price/m2) of "Valdemorillo" and "Galapagar" the same? (★★☆)

Print both average prices and then write a conclusion about it.

Hint: Create a new column called `pps` (price per square meter) and then analyze the values.

```python
# TODO

ds = pd.read_csv('assets/real_estate.csv', sep = ';')

# Selecciono las filas que concuerdan con 'Valdemorillo' y 'Galapagar'
ds_filtered_pps = ds[ds['level5'].isin(['Valdemorillo', 'Galapagar'])]

# Creo la columna 'pps' con los valores de price/surface
ds['pps'] = ds_filtered_pps['price'] / ds_filtered_pps['surface']

square_meter_Valdemorillo = ds[ds['level5'] == 'Valdemorillo']['pps'].mean()
square_meter_Galapagar = ds[ds['level5'] == 'Galapagar']['pps'].mean()

print(f"El promedio del precio por metro cuadrado en Valdemorillo es de : {round(square_meter_Valdemorillo, 2)} EUR/m^2\n"
      f"El promedio del precio por metro cuadrado en Galapagar es de: {round(square_meter_Galapagar, 2)} EUR/m^2.")

diferencia_absoluta = square_meter_Galapagar - square_meter_Valdemorillo
diferencia_relativa = ((square_meter_Galapagar - square_meter_Valdemorillo) / square_meter_Valdemorillo) * 100

print(f"Esto representa una diferencia absoluta de {round(diferencia_absoluta,2)} EUR por metro cuadrado.\n"
      f"Relativamente, el precio en Galapagar es aproximadamente un {round(diferencia_relativa,2)}% más alto que en Valdemorillo.")
```

#### Exercise 15. Make a bar plot of the median of the prices and explain what you observe (you must use the subset obtained in Exercise 14) (★★★)

Print the bar of the median of the prices and write in the Markdown cell a brief analysis about the plot.

```python
# TODO: Code

ds_filtered = ds[ds['price'] != 0]

south_belt = ds_filtered[ds_filtered['level5'].isin(["Fuenlabrada","Leganés","Getafe","Alcorcón"])]

#Agrupa las filas del DataFrame south_belt por el valor de la columna level5 
# (es decir, por municipio: Fuenlabrada, Leganés, Getafe, Alcorcón).

# Dentro de cada grupo, selecciona solo la columna price.

# Calcula la mediana de los precios dentro de cada grupo.
median_price = south_belt.groupby('level5')['price'].median()
median_price

# Gráfico de barras 
plt.bar( x = median_price.index, height = median_price.values,
        edgecolor = 'black',
        color = 'darkblue',
        alpha = 0.9)

# Título y labels
plt.title('Mediana de los precios')
plt.xlabel('Población')
plt.ylabel('Precio')

#Forma optimizada con textos dinámicos
# value + 5000(valor cualquiera) es para que el valor no se superponga con la barra. 
for index, value in enumerate(median_price.values):
    plt.text(index, value + 5000, f'{round(value, 2)}', ha='center', fontsize=10, color='black')

plt.show()
```

#### Exercise 18. Normalize the variable of prices for each population and plot the 4 histograms in the same plot (you must use the subset obtained in Exercise 14) (★★★)

For the normalization method, you can use the one you consider; there is not a single correct answer to this question. Print the plot and write in the Markdown cell a brief analysis about the plot.

Hint: You can help yourself by reviewing the *multihist* demo of Matplotlib.

```python
# TODO

import pandas as pd
import matplotlib.pyplot as plt

pd.options.mode.chained_assignment = None  # default='warn'

ds_filtered = ds[ds['price'] != 0]

south_belt = ds_filtered[ds_filtered['level5'].isin(["Fuenlabrada","Leganés","Getafe","Alcorcón"])]

# Column with normalize prices

south_belt['Normalize'] = (south_belt['price'] - south_belt['price'].mean()) / south_belt['price'].std()


# Normalize values for each city

normalize_fuenlabrada = south_belt.loc[south_belt['level5'] == 'Fuenlabrada', 'Normalize']
normalize_leganes = south_belt.loc[south_belt['level5'] == 'Leganés', 'Normalize']
normalize_getafe = south_belt.loc[south_belt['level5'] == 'Getafe', 'Normalize']
normalize_alcorcon = south_belt.loc[south_belt['level5'] == 'Alcorcón', 'Normalize']

# Figure and subs

fig, axis = plt.subplots(2,2, figsize = (10,8))

# Histograma Fuenlabrada
axis[0, 0].hist(normalize_fuenlabrada, bins=30, color='blue', edgecolor='black')
axis[0, 0].set_title('Fuenlabrada')

# Histograma Leganés
axis[0, 1].hist(normalize_leganes, bins=30, color='green', edgecolor='black')
axis[0, 1].set_title('Leganés')

# Histograma Getafe
axis[1, 0].hist(normalize_getafe, bins=30, color='red', edgecolor='black')
axis[1, 0].set_title('Getafe')

# Histograma Alcorcón
axis[1, 1].hist(normalize_alcorcon, bins=30, color='purple', edgecolor='black')
axis[1, 1].set_title('Alcorcón')

# Ajustar el diseño
plt.tight_layout()


plt.show()
```
